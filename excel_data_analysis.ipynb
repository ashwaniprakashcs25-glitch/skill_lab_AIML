```python
from sklearn.model_selection import train_test_split #creating model idk how+
x= df.drop('Sales',axis=1)
y=df['Sales']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train_selected = x_train
x_test_selected = x_test

print("selected features:",x_train_selected.columns.tolist())
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train_selected,y_train)
```

```python
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error # linear regression idk how
y_pred=model.predict(x_test_selected)
r2=r2_score(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test, y_pred)
print("R2 Score:",r2)
print("MSE:",mse)
print("MAE:",mae)
```
