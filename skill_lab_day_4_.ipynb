```python
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.datasets import load_iris
import seaborn as sns
from sklearn.model_selection import train_test_split
iris=load_iris()
```

```python
print('dataset description\n:',iris.DESCR) # OR #iris['descr']
print('type of object returned',type(load_iris()))
```

```python
type(iris.data)
```

```python
print('keys of dataset',iris.keys())
print('shape of target key ',iris.target.shape)
```

```python
iris_df=pd.DataFrame(data=iris.data,columns=iris.feature_names)
iris_df['target']=iris.target #add new cols
print(iris_df.head())
```

```python
from sklearn.model_selection import train_test_split #tts also work for unsupervised ml
ytrain,ytest,xtrain,xtest=train_test_split(iris.target,iris.data,test_size=0.2)# the test size parameters define the testing data percentage
```

```python
iris.data.shape
```

```python
print('training output shape ',ytrain.shape)
print('training input shape ',xtrain.shape)
print('test input shape',xtest.shape)
print('test output shape',ytest.shape)
```

```python
plt.figure(figsize=(15,8))
# subplot 1 (for supervised data
plt.subplot(1,2,1)#1 row 2 col
plt.scatter(xtrain[:,2],xtrain[:,3],cmap='magma',c='blue',edgecolors='k')# includes only training data "#plt.scatter(iris.data[:,2],iris.data[:,3])# includes whole table train and test both", cmap is for boundary and c stands for color parameter
plt.grid()
plt.xlabel('petal length')
plt.ylabel('petal width')
plt.title('supervised  data')
#subplot 2
plt.subplot(1,2,2)
plt.scatter(xtest[:,2],xtest[:,3],edgecolors='k',cmap='magma',c='yellow')# includes only testing data
plt.grid()
plt.xlabel('petal length')
plt.ylabel('petal width')
plt.title('unsupervised  data')

```

```python
from sklearn.cluster import KMeans # kmeans is a class any mod with capital 1st letter is class
Kmeans=KMeans(n_clusters=3)# build model (object) we have 3 eqn an 4 VAR
Kmeans.fit(xtrain)#model training
train_labels=Kmeans.predict(xtrain)# train
test_labels=Kmeans.predict(xtest)# test
```

```python
from sklearn.metrics import silhouette_score
print('silhouette score for training data',silhouette_score(xtrain,train_labels))
print('silhouette score for testing data',silhouette_score(xtest,test_labels))
```

```python
#subplot 3
plt.subplot(1,3,3)
plt.scatter()# includes only testing data
plt.grid()
plt.xlabel('petal length')
plt.ylabel('petal width')
plt.title('unsupervised  data')


plt.tight_layout()
plt.show()
```
