```python
from sklearn.model_selection import train_test_split #tts also work for unsupervised ml
ytrain,ytest,xtrain,xtest=train_test_split(iris.target,iris.data)
print('type of training inputs ',type(xtrain))
print('type of training outputs ',type(ytrain))
print('type of testing inputs ',type(xtest))
print('type of testing outputs ',type(ytest))
```

```python
from sklearn.model_selection import train_test_split #tts also work for unsupervised ml
ytrain,ytest,xtrain,xtest=train_test_split(iris_df['target'],iris_df.drop('target',axis=1))
print('type of training inputs ',type(xtrain))
print('type of training outputs ',type(ytrain))
print('type of testing inputs ',type(xtest))
print('type of testing outputs ',type(ytest))
```

```python
from sklearn.linear_model import LogisticRegression
lReg=LogisticRegression()#build model object
lReg.fit(xtrain,ytrain)#model training -->m and c
lReg.coef_
print('shape of coefficient:,1Reg.coef_.shape')#(class,features)
print('coefficient (m):\n',lReg.coef_)
print('shape of intercept:,1Reg.intercept_.shape')#(class)
print('intercept (c):\n',lReg.intercept_)
```

```python
print(' training accuracy score:',lReg.score(xtest,ytest))
print(' testing accuracy score:',lReg.score(xtrain,ytrain))
```

```python
#best feature - "petal width(cm)"
lReg=LogisticRegression()#build model object
lReg.fit(xtrain[['petal width (cm)']],ytrain)#model training -->m and c
print(' training accuracy score:',lReg.score(xtest[['petal width (cm)']],ytest))
print(' testing accuracy score:',lReg.score(xtrain[['petal width (cm)']],ytrain))
```
