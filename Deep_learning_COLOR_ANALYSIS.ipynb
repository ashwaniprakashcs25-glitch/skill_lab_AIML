```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from  PIL import Image
import cv2
```

```python
image=Image.open(r'/content/8X8pixel.png')#load image
print('type of object returned:',type(image))
```

```python
#plot image
image#or image.show(# for non notebook env)
plt.imshow(image)
plt.axis('off')
plt.show()#non notebook environment
```

```python
image_grayscale=image.convert('L')#to map out black and white)#GRAYSCALE
#plot grayscale img
plt.imshow(image_grayscale,cmap='gray')
plt.axis('off')
plt.show()
```

```python
grayscale_image_array=np.array(image_grayscale)
print('type of object returned:',type(grayscale_image_array))
print('shape of array:',grayscale_image_array.shape)
print(grayscale_image_array)
print('min value:',np.min(grayscale_image_array))
print('max value:',np.max(grayscale_image_array))
print(f"pixel range:",({grayscale_image_array.min()},f"{grayscale_image_array.max()}"))
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(grayscale_image_array,cmap='gray')
plt.axis('off')
plt.title('grayscale_image_array')
plt.subplot(1,2,2)
plt.imshow(image_grayscale,cmap='gray')
plt.axis('off')
plt.title('grayscale image')
sns.heatmap(grayscale_image_array,cmap='gray',annot=True,fmt='',cbar=False)#annot stands for annotation true will give value of color , fmt is for format by default it give scientific notation to normal

plt.tight_layout()
plt.show()
```

```python
image_array=np.array(image)
#image_array=cv2.imread('address,1)
#where
#1 or cv2.IMREAD_COLOR
#0 or cv2.IMREAD_GRAYSCALE
#-1 or cv2.IMREAD_UNCHANGED
print('type of object returned:',type(image_array))
print('shape of array:',image_array.shape)#(8,8,4) 4 for transparency 0-r,1-g,2-b
print(image_array)
print('min value:',np.min(image_array))
print('max value:',np.max(image_array))

print(f"pixel range:",({image_array.min()},f"{image_array.max()}"))
```

```python
plt.figure(figsize=(20,4))
plt.subplot(1,4,1)
plt.imshow(image_array,cmap='gray')
plt.axis('off')
plt.title('original image')


plt.subplot(1,4,2)
plt.axis('off')
plt.title(' red channel')
sns.heatmap(image_array[:,:,0],cmap='Reds',annot=True,fmt='',cbar=False)#annot stands for annotation true will give value of color , fmt is for format by default it give scientific notation to normal

plt.subplot(1,4,3)
plt.axis('off')
plt.title(' green channel')
sns.heatmap(image_array[:,:,1],cmap='Greens',annot=True,fmt='',cbar=False)

plt.subplot(1,4,4)
plt.axis('off')
plt.title(' blue channel')
sns.heatmap(image_array[:,:,2],cmap='Blues',annot=True,fmt='',cbar=False)

plt.show()
```
