```python
list2d = [ [2,5,7,69],  #row1
           [3,4,6,70],  #row2
           [1,0,0,0] ] #row3
import numpy as np
array2d = np.array(list2d,dtype = 'int16')
print('return type',type(array2d))
print('type of data in array',array2d.dtype) # tells type of data present in array int64 bit is from c++ environment
print(array2d)
print('dims:',array2d.ndim)#no of dims (dimension 1D)
print('shape of array',array2d.shape)#shape of array output is a  tuple with each value representing each axis
print('size of array',array2d.size)#size of array
print('sum of all values',array2d.sum())
print('mean of all values',array2d.mean())
print('maximum value in array',array2d.max())
print('minimum value in array',array2d.min())
print('variance of all values',array2d.var())
print('standard deviation of all values',array2d.std())
print( 'maximum value in array row1 ',array2d[0].max())# the axis is assigned by no index starting from zero and goes right to left
print( 'maximum value in array row2 ',array2d[1].max())
print( 'maximum value in array row3 ',array2d[2].max())
print( 'maximum value in array col1 ',array2d[:,0].max())
print( 'maximum value in array col2 ',array2d[:,1].max())
print( 'maximum value in array col3 ',array2d[:,2].max())
print( 'maximum value in array col4 ',array2d[:,3].max())
print( 'minimum value in array col1 ',array2d[:,0].min())
print( 'minimum value in array col2 ',array2d[:,1].min())
print( 'minimum value in array col3 ',array2d[:,2].min())
print( 'minimum value in array col4 ',array2d[:,3].min())
print( 'maximum value in array columns',array2d.max(axis=1))
print( 'minimum value in array columns',array2d.min(axis=1))
print( 'minimum value in array rows',array2d.min(axis=0))
print( 'maximum value in array rows',array2d.max(axis=0))

```

```python
# maths array* array
array_a= np.array([[1,2,3],[1,2,3]],dtype='int64') # 2 rows and 3 columns
array_b = np.array([[4,5,6],[6,7,8]],dtype='int64')
print('array a is ',array_a , end='\n\n\n',sep="")
print('array b is ',array_b,'\n')
print('sum of array is',array_a + array_b,'\n\n')#np.add(array_a,array_b)
print(array_a - array_b)# again all elements are elementary operation basis
print(array_a * array_b)#
print(array_a / array_b)
print(array_a // array_b)
print(array_a % array_b)
print(array_a ** array_b)
```

```python
#broad casting (same concept of type casting and promotion the smaller matrix is promoted to higher size to conclude byu not the same way )
array_c=np.array([[-1,0,-1]])#shape(1,3)
#IMPORTANT
print('add a and c:\n',array_a + array_c)# see first row 1 of a added into b then row 2 of a added to b
print('sub a and c:\n',array_a - array_c)
print('mult a and c:\n',array_a * array_c)#same row1 to b then row 2 to b
```

```python
#vector products
dot= np.dot(array_a,array_b)#dot product
cross= np.cross(array_a,array_b)#cross product
print('dot product is',dot)
print('cross product is',cross)
```

```python
#matrix multiplication
p = array_a @ array_b.transpose() # matrix multiplication or you can use np.matmul(array_a,array_b)
print(p)
```

```python

#np.random.normal(loc=,scale=) is used to make a continuous distribution loc--location of max value scale-- standard deviatiion
blue=np.random.normal(loc=0.0,scale=0.2,size=1000)
black=np.random.normal(loc=0.0,scale=1.0,size=1000)
green=np.random.normal(loc=0.0,scale=5.0,size=1000)
purple=np.random.normal(loc=-2.0,scale=0.5,size=1000)

plt.hist(blue, color='blue', histtype='step', density=True, label='loc=0.0, scale=0.2 ')
plt.hist(black, color='black', histtype='step', density=True, label='loc=0.0, scale=1.0 ')
plt.hist(green, color='green', histtype='step', density=True, label='loc=0.0, scale=5.0 ')
plt.hist(purple, color='purple', histtype='step', density=True, label='loc=-2.0, scale=1.0 ')


plt.grid()
plt.title('Random Probability Distributions ')
plt.xlabel('Values')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```

```python


#np.random.normal(loc=,scale=) is used to make a continuous distribution loc--location of max value scale-- standard deviatiion
blue=np.random.normal(loc=0.0,scale=0.2,size=1000)
black=np.random.normal(loc=0.0,scale=1.0,size=1000)
green=np.random.normal(loc=0.0,scale=5.0,size=1000)
purple=np.random.normal(loc=-2.0,scale=0.5,size=1000)
#stat --> 'count','frequency','probability(or proportion)' 'percent','density'
#element ->'bars','step','poly'

sns.histplot(blue, color='blue',stat='probability',element='poly',label='loc=0.0, scale=0.2',alpha=0.5) #alpha==transparency parameter
sns.histplot(black, color='black',stat='probability',element='poly', label='loc=0.0, scale=1.0',alpha=0.4)
sns.histplot(green, color='green', stat='probability',element='poly', label='loc=0.0, scale=5.0 ',alpha=0.3)
sns.histplot(purple, color='purple', stat='probability',element='poly', label='loc=-2.0, scale=1.0',alpha=0.2)

plt.grid()
plt.title('Random Probability Distributions ')
plt.xlabel('Values')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
