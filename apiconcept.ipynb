```python
import openai
from google.colab import userdata
```

```python
client=openai.OpenAI(api_key=userdata.get('openaikey'))
```

```python
response=client.responses.create(model='gpt-5.6-luna',input='who are you?')
print('type of output:',type (response))
```

```python
print(response.output_text)
```
