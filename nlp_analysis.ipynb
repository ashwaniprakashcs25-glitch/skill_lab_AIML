```python
# step 1 : load the training data (only these four categories )
data_train=fetch_20newsgroups(subset='train',categories=categories,shuffle=True,random_state=42)
```

```python
# step 2 :load the test data
data_test=fetch_20newsgroups(subset='test',categories=categories,shuffle=True,random_state=42)
```

```python
# step3 : get the category name and labels (targets)
target_names=data_train.target_names
y_train = data_train.target
y_test = data_test.target
```

```python
#step 4 : calculate the size of data in mb (just for info )
train_size_mb=sum(len(doc.encode('utf-8')) for doc in data_train.data)/ 1e6
test_size_mb=sum(len(doc.encode('utf-8')) for doc in data_test.data)/ 1e6
```

```python
#fit on training data and transform it into vectors
t0=time()
x_train = vectorizer.fit_transform(data_train.data)
duration=time()-t0
```
